#DATA CLEANING

##Data Cleaning

The datasets were cleaned before analysis by standardizing column names, removing duplicate records, handling missing values, and verifying the data types. These preprocessing steps improve data quality and ensure reliable analytical results.

##Create Copies of the Original Datasets

```python
# ==========================================================
# STEP 12: Create Working Copies
# ==========================================================

coverage = coverage_df.copy()
incidence = incidence_df.copy()
reported = reported_df.copy()
introduction = intro_df.copy()
schedule = schedule_df.copy()

print("Working copies created successfully.")
```

##Standardize Column Names

```python
# ==========================================================
# STEP 13: Standardize Column Names
# ==========================================================

def clean_column_names(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.replace("/", "_")
    )
    return df

coverage = clean_column_names(coverage)
incidence = clean_column_names(incidence)
reported = clean_column_names(reported)
introduction = clean_column_names(introduction)
schedule = clean_column_names(schedule)

print("Column names standardized successfully.")
```

##Remove Duplicate Records

```python
# ==========================================================
# STEP 14: Remove Duplicate Records
# ==========================================================

datasets = {
    "Coverage": coverage,
    "Incidence": incidence,
    "Reported": reported,
    "Introduction": introduction,
    "Schedule": schedule
}

for name, df in datasets.items():
    before = df.shape[0]
    df.drop_duplicates(inplace=True)
    after = df.shape[0]
    print(f"{name}: Removed {before - after} duplicate rows")
```

##Check Missing Values Again

```python
# ==========================================================
# STEP 15: Missing Value Summary
# ==========================================================

for name, df in datasets.items():
    print(f"\n{name} Dataset")
    display(df.isnull().sum())
```

##Handle Missing Values

```python
# ==========================================================
# STEP 16: Handle Missing Values
# ==========================================================

for name, df in datasets.items():

    # Fill numeric columns with median
    numeric_cols = df.select_dtypes(include=np.number).columns
    df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

    # Fill categorical columns with mode
    categorical_cols = df.select_dtypes(include="object").columns

    for col in categorical_cols:
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled successfully.")
```

##Verify Missing Values

```python
# ==========================================================
# STEP 17: Verify Missing Values
# ==========================================================

for name, df in datasets.items():
    print(f"{name}: {df.isnull().sum().sum()} missing values remaining")
```

#SQL DATABASE SETUP

```python
# ==========================================================
# SQL DATABASE SETUP
# Store cleaned vaccination datasets into relational tables
# ==========================================================

import sqlite3

conn = sqlite3.connect("vaccination_database.db")
cursor = conn.cursor()

print("Database created successfully.")
```

##Normalised Tables

```python
# ==========================================================
# Create Dimension Tables
# ==========================================================

cursor.execute("""
CREATE TABLE IF NOT EXISTS countries (
    country_id INTEGER PRIMARY KEY AUTOINCREMENT,
    code TEXT,
    country_name TEXT UNIQUE
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS years (
    year_id INTEGER PRIMARY KEY AUTOINCREMENT,
    year INTEGER UNIQUE
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS vaccines (
    vaccine_id INTEGER PRIMARY KEY AUTOINCREMENT,
    vaccine_code TEXT,
    vaccine_description TEXT UNIQUE
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS diseases (
    disease_id INTEGER PRIMARY KEY AUTOINCREMENT,
    disease_code TEXT,
    disease_description TEXT UNIQUE
)
""")

conn.commit()

print("Dimension tables created successfully.")
```

##Insertion of dimension data

```python
# ==========================================================
# Insert Countries
# ==========================================================

countries = coverage[["code", "name"]].copy()
countries.columns = ["code", "country_name"]
countries = countries.drop_duplicates(subset=["country_name"])

countries.to_sql("countries", conn, if_exists="append", index=False)

print("Countries inserted successfully.")
```

```python
# ==========================================================
# Insert Years
# ==========================================================

years = pd.DataFrame({
    "year": sorted(coverage["year"].dropna().unique())
})

years.to_sql("years", conn, if_exists="append", index=False)

print("Years inserted successfully.")
```

```python
# ==========================================================
# Insert Vaccines
# ==========================================================

vaccines = coverage[["antigen", "antigen_description"]].drop_duplicates()
vaccines.columns = ["vaccine_code", "vaccine_description"]

vaccines.to_sql("vaccines", conn, if_exists="append", index=False)

print("Vaccines inserted successfully.")
```

```python
# ==========================================================
# Insert Diseases
# ==========================================================

diseases = incidence[["disease", "disease_description"]].drop_duplicates()
diseases.columns = ["disease_code", "disease_description"]

diseases.to_sql("diseases", conn, if_exists="append", index=False)

print("Diseases inserted successfully.")
```

##Fact Tables

```python
# ==========================================================
# Create Vaccination Coverage Fact Table
# ==========================================================

cursor.execute("""
CREATE TABLE IF NOT EXISTS vaccination_coverage (
    coverage_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_name TEXT,
    year INTEGER,
    vaccine_description TEXT,
    coverage_category TEXT,
    target_number REAL,
    doses REAL,
    coverage REAL,
    FOREIGN KEY(country_name) REFERENCES countries(country_name),
    FOREIGN KEY(year) REFERENCES years(year),
    FOREIGN KEY(vaccine_description) REFERENCES vaccines(vaccine_description)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS disease_incidence (
    incidence_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_name TEXT,
    year INTEGER,
    disease_description TEXT,
    denominator REAL,
    incidence_rate REAL,
    FOREIGN KEY(country_name) REFERENCES countries(country_name),
    FOREIGN KEY(year) REFERENCES years(year),
    FOREIGN KEY(disease_description) REFERENCES diseases(disease_description)
)
""")

conn.commit()

print("Fact tables created successfully.")
```

##Insertion of fact data

```python
# ==========================================================
# Insert Vaccination Coverage Data
# ==========================================================

vaccination_fact = coverage[[
    "name",
    "year",
    "antigen_description",
    "coverage_category",
    "target_number",
    "doses",
    "coverage"
]].copy()

vaccination_fact.columns = [
    "country_name",
    "year",
    "vaccine_description",
    "coverage_category",
    "target_number",
    "doses",
    "coverage"
]

vaccination_fact.to_sql(
    "vaccination_coverage",
    conn,
    if_exists="append",
    index=False
)

print("Vaccination coverage data inserted successfully.")
```

```python
# ==========================================================
# Insert Disease Incidence Data
# ==========================================================

incidence_fact = incidence[[
    "name",
    "year",
    "disease_description",
    "denominator",
    "incidence_rate"
]].copy()

incidence_fact.columns = [
    "country_name",
    "year",
    "disease_description",
    "denominator",
    "incidence_rate"
]

incidence_fact.to_sql(
    "disease_incidence",
    conn,
    if_exists="append",
    index=False
)

print("Disease incidence data inserted successfully.")
```

##SQL QUERIES

```python
# ==========================================================
# Query 1: Top 10 Countries by Average Vaccination Coverage
# ==========================================================

query = """
SELECT country_name, AVG(coverage) AS avg_coverage
FROM vaccination_coverage
GROUP BY country_name
ORDER BY avg_coverage DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# ==========================================================
# Query 2: Top Diseases by Incidence Rate
# ==========================================================

query = """
SELECT disease_description, AVG(incidence_rate) AS avg_incidence
FROM disease_incidence
GROUP BY disease_description
ORDER BY avg_incidence DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# ==========================================================
# Query 3: Year-wise Average Vaccination Coverage
# ==========================================================

query = """
SELECT year, AVG(coverage) AS avg_coverage
FROM vaccination_coverage
GROUP BY year
ORDER BY year;
"""

pd.read_sql_query(query, conn)
```

### SQL Database Setup

The cleaned vaccination datasets were stored in a relational SQLite database. The database was normalized into separate dimension tables such as countries, years, vaccines, and diseases to reduce redundancy.

Fact tables were created for vaccination coverage and disease incidence data. Primary keys were used in dimension tables, and foreign keys were used in fact tables to maintain referential integrity.

This structure improves data organization, makes querying easier, and supports efficient analysis using SQL.
